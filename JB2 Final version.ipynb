# Data understanding

To calculate air quality, we need the following values: CO2 and humidity.
We can then calculate the air quality as follows:

#### Carbon Dioxide (CO2):
- Carbon dioxide is an indicator of ventilation efficiency and the presence of polluted air in a room. Higher levels of CO2 can indicate insufficient ventilation and an accumulation of pollutants.
- The concentration of CO2 is often measured in parts per million (ppm). Guidelines for the maximum acceptable concentration of CO2 vary, but generally, a level of 1000 ppm is considered acceptable for good indoor air quality.
- Higher levels of CO2 can lead to fatigue, headaches, concentration problems, and reduced productivity among occupants.
#### Humidity:
- Humidity is an important factor for the comfort and health of occupants. Too high or too low humidity levels can cause various problems, such as respiratory irritation, dry skin, condensation issues, and mold growth.
- The ideal humidity for indoor environments is usually between 40% and 60%. This level provides a comfortable indoor climate and minimizes the risk of moisture-related problems.
- Humidity is typically measured in percent relative humidity (%RH).

#### Berekenen daily difference

```python
# voor de data source remeha
def daily_difference_remeha(df, id_list):
    df_final = pd.DataFrame()
    df = df.reset_index()
    for id_val in id_list:
        df_temp = df[df['id'] == id_val].copy()
        df_temp.loc[:, 'g_use_ch__kWh'] = df_temp['g_use_ch_inf_cum__kWh'].diff().fillna(0)
        df_temp.loc[:, 'g_use_dhw__kWh'] = df_temp['g_use_dhw_inf_cum__kWh'].diff().fillna(0)
        df_final = pd.concat([df_final, df_temp], ignore_index=True)
    df_final = df_final.drop(columns=['g_use_ch_inf_cum__kWh', 'g_use_dhw_inf_cum__kWh'])
    return df_final  

# voor de data source Need For Heat
def daily_difference_nfh(df, id_list):
    df_final = pd.DataFrame()
    df = df.reset_index()
    for id_val in id_list:
        df_temp = df[df['id'] == id_val].copy()
        df_temp.loc[:, 'g_use__m3'] = df_temp['g_use_cum__m3'].diff().fillna(0)
        df_final = pd.concat([df_final, df_temp], ignore_index=True)
    df_final = df_final.drop(columns=['g_use_cum__m3'])
    return df_final  

# voor de later toegevoegde data
def daily_difference_nfh_toevoeging(df, id_list):
    df_final = pd.DataFrame()
    df = df.reset_index()
    for id_val in id_list:
        df_temp = df[df['id'] == id_val].copy()
        df_temp.loc[:, 'rel_humidity__0'] = df_temp['rel_humidity__0'].diff().fillna(0)
        df_temp.loc[:, 'co2__ppm'] = df_temp['co2__ppm'].diff().fillna(0)
        df_temp.loc[:, 'H__W_K_1'] = df_temp['H__W_K_1'].diff().fillna(0)
        df_final = pd.concat([df_final, df_temp], ignore_index=True)
    return df_final  
```

```python
def plot_linear_regression_by_ids(df, id_list):
    for unique_id in id_list:
        # Select data for the current ID
        df_filtered = df[df['id'] == unique_id].copy()
        
        # Check if there is data for the current ID
        if df_filtered.empty:
            print(f"No data for ID {unique_id}.")
            continue
        
        # Convert data to appropriate data types
        df_filtered['rel_humidity__0'] = df_filtered['rel_humidity__0'].astype(float)
        df_filtered['H__W_K_1'] = df_filtered['H__W_K_1'].astype(float)
        df_filtered['co2__ppm'] = df_filtered['co2__ppm'].astype(float)
        
        # Build the regression model for 'H__W_K_1' and 'humidity (%)'
        X_humidity = np.asarray(df_filtered['rel_humidity__0'])
        y = np.asarray(df_filtered['H__W_K_1'])
        model_humidity = sm.OLS(y, sm.add_constant(X_humidity)).fit()
        r_squared_humidity = model_humidity.rsquared
        
        # Build the regression model for 'H__W_K_1' and 'Co2 (ppm)'
        X_co2 = np.asarray(df_filtered['co2__ppm'])
        model_co2 = sm.OLS(y, sm.add_constant(X_co2)).fit()
        r_squared_co2 = model_co2.rsquared
        
        # Plot the graphs
        plt.figure(figsize=(12, 6))
        
        # Plot 'H__W_K_1' and 'humidity (%)'
        plt.subplot(1, 2, 1)
        plt.scatter(X_humidity, y, color='blue', label='Original data')
        plt.plot(X_humidity, model_humidity.predict(sm.add_constant(X_humidity)), color='red', label='Linear regression')
        plt.xlabel('rel_humidity__0')
        plt.ylabel('H__W_K_1')
        plt.title(f'Linear Regression for ID {unique_id} (humidity)\nR-squared: {r_squared_humidity:.2f}')
        plt.legend()
        
        # Plot 'H__W_K_1' and 'Co2 (ppm)'
        plt.subplot(1, 2, 2)
        plt.scatter(X_co2, y, color='blue', label='Original data')
        plt.plot(X_co2, model_co2.predict(sm.add_constant(X_co2)), color='red', label='Linear regression')
        plt.xlabel('co2__ppm')
        plt.ylabel('H__W_K_1')
        plt.title(f'Linear Regression for ID {unique_id} (Co2)\nR-squared: {r_squared_co2:.2f}')
        plt.legend()
        
        plt.tight_layout()
        plt.show()
```

#### Functie voor het maken van een label voor de lucht kwaliteit

```python
def label_air_quality_with_id(df):
    humidity_percentages = df['rel_humidity__0'] * 100
    # Definieer de grenswaarden voor CO2-concentratie en luchtvochtigheid
    co2_quality = pd.cut(df['co2__ppm'], bins=[-float('inf'), 400, 800, 1200, float('inf')], labels=['Very Poor', 'Poor', 'Moderate', 'Good'], right=False)
    humidity_quality = pd.cut(humidity_percentages, bins=[-float('inf'), 20, 30, 60, 70, float('inf')], labels=['Very Poor', 'Poor', 'Moderate', 'Good', 'Very Good'], right=False)
    
    # Combineer de luchtkwaliteit op basis van CO2 en luchtvochtigheid
    overall_quality = pd.Series(index=co2_quality.index)
    overall_quality[(co2_quality == 'Good') & (humidity_quality == 'Good')] = 'Good'
    overall_quality[~((co2_quality == 'Good') & (humidity_quality == 'Good'))] = 'Moderate'
    overall_quality[(co2_quality == 'Very Poor') | (humidity_quality == 'Very Poor')] = 'Poor'
    
    # Maak een DataFrame met de ID's en gelabelde luchtkwaliteit
    result_df = pd.DataFrame({
        'ID': df['id'],
        'Air Quality': overall_quality
    })
    
    return result_df
```
