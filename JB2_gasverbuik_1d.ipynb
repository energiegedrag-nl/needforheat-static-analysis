# Data understanding

To calculate the air quality, we need the following value: CO2 and humidity. 
We can then calculate the air quality in the following way: <br>
#### Carbon dioxide (CO2):

<li>Carbon dioxide is an indicator of ventilation efficiency and the presence of polluted air in a room. Higher levels of CO2 may indicate inadequate ventilation and a buildup of pollutants.
<li>The concentration of CO2 is often measured in parts per million (ppm). Guidelines for the maximum acceptable concentration of CO2 vary, but in general, a level of 1000 ppm is considered acceptable for good indoor air quality.
<li>Higher levels of CO2 can lead to fatigue, headaches, difficulty concentrating, and reduced productivity in residents.

#### Humidity:

<li>Humidity is an important factor for the comfort and health of residents. Humidity levels that are too high or too low can cause a variety of problems, such as respiratory irritation, dry skin, condensation problems, and mold growth.
<li>The ideal humidity for indoor environments is usually between 40% and 60%. This level provides a comfortable indoor climate and minimizes the risk of moisture-related problems.
<li>Humidity is usually measured in relative humidity (%RH).

```python
#Separating the column as example code
data = df_house_id['Gas usage cum ch (kWh)']

#Function to decumulate data
def decumulate(data):
    non_cumulative = [data.iloc[0]]  
    for i in range(1, len(data)):
        non_cumulative.append(data.iloc[i] - data.iloc[i - 1])
    return non_cumulative
```

```python
#Decumulating data
df_house_id['Gas usage ch (kWh)'] = decumulate(df_house_id['Gas usage cum ch (kWh)'])
```

### Evaluation
The question being answered with the models provided in this notebook is: Can you predict gas-usage based on inside temperature (and outside temperature)? According to the heatmaps done and the linear regression model, the inside temperature and gas usage for central heating don't have enough correlation to optimally predict the gas usage.

</br>

#### evaluation criteria
For the results we look at both R^2 values of the original train/test split and the train/test split with the random state (_42_) + descriptive heatmap model for the _mean_diff_rem_house_id_.
